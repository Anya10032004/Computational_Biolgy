Sequence A:

AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC

Sequence B:

CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG


Please use Biopython to:

 •	Find the length of both sequences.

 •	Determine the number of times the codon (CGC) appears in both sequences.

 •	Locate the first occurrence index position of the sub-sequence (CAGTC) in both sequences.

 •	Combine the first 15 nucleotides from sequence A and the last 10 nucleotides from sequence B to create a new DNA sequence called sequence C.

 •	Reverse the sequence of sequence C.

 •	Plot the nucleotide base (A, C, G, T) frequency of each of the sequence A, sequence B, and sequence C using the Matplotlib library.

Bagian 1

```python
import Bio
from Bio.Seq import Seq
seqA = Seq("AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC")
seqB = Seq("CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG")

print("Sequence A length:", len(seqA))
print("Sequence B length:", len(seqB))
```

Bagian 2

```python
from Bio.SeqUtils import nt_search
subseq = Seq("CGC")
seqA_subseq = nt_search(str(seqA), str(subseq))
seqB_subseq = nt_search(str(seqB), str(subseq))

print(seqA_subseq)
print(seqB_subseq)

print("The number of Codon CGC appears in sequence A is ", len(seqA_subseq) - 1)
print("The number of Codon CGC appears in sequence B is ", len(seqB_subseq) - 1)
```

Bagian  3

```python
def find_index(indexss, nameSeq):
    if len(indexss) == 1:
        print(f"{indexss[0]} is not found in {nameSeq}")
        
    for a in range(0, len(indexss)):
        if a == 0:
            continue
        print(f"{indexss[0]} is found in {nameSeq} at index {indexss[a]}")

subseq2 = Seq("CAGTC")
seqA_subseq2 = nt_search(str(seqA), str(subseq2))
seqB_subseq2 = nt_search(str(seqB), str(subseq2))

print(seqA_subseq2)
print(seqB_subseq2)

find_index(seqA_subseq2, "Sequence A")
find_index(seqB_subseq2, "Sequence B")
```

Bagian 4

```python
seqAA = seqA[0:15]
print(len(seqAA))

seqBB = seqB[(len(seqB) - 10):len(seqB)]
print(len(seqBB))

seqC = seqAA + seqBB
print("So Sequence C is", seqC)
```

Bagian 5

```python
# Reverse of sequqnce C
print("The reverse of Sequence C is", seqC[len(seqC)-1::-1])
```

Bagian 6

```python
from collections import Counter
dna_freq_seqA = Counter(seqA)
print("DNA frequency in Sequence A:", dna_freq_seqA)

dna_freq_seqB = Counter(seqB)
print("DNA frequency in Sequence B:", dna_freq_seqB)

dna_freq_seqC = Counter(seqC)
print("DNA frequency in Sequence C:", dna_freq_seqC)

```

```python
from matplotlib import pyplot as plt    
plt.figure(figsize=(8, 6))
plt.bar(dna_freq_seqA.keys(), dna_freq_seqA.values())
plt.xlabel("Nucleotide")
plt.ylabel("Frequency")
plt.title("DNA Frequency in Sequence A")
plt.show()

plt.figure(figsize=(8, 6))
plt.bar(dna_freq_seqB.keys(), dna_freq_seqB.values())
plt.xlabel("Nucleotide")
plt.ylabel("Frequency")
plt.title("DNA Frequency in Sequence B")
plt.show()
plt.figure(figsize=(8, 6))
plt.bar(dna_freq_seqC.keys(), dna_freq_seqC.values())
plt.xlabel("Nucleotide")
plt.ylabel("Frequency")
plt.title("DNA Frequency in Sequence C")
plt.show()

```
