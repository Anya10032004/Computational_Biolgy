```python
file_name = 'sequences.fasta'
for record in SeqIO.parse(file_name, 'fasta'):
    print(record)
    print('------------------------')
    print("ID              : ", record.id)
    print("name            : ", record.name)
    print("Description     : ", record.description)
    print("Sequence        : ", record.seq)
```
