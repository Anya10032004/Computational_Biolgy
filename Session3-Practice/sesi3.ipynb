Please use Biopython to:

•	Transcribe the DNA sequence A and sequence B into mRNA.

•	Combine 24 starting nucleotides of mRNA sequence A and 21 ending nucleotides of mRNA sequence B and set it as mRNA sequence C.

•	Translate the mRNA sequence A, sequence B, and sequence C into amino acids sequences.

•	Transcribe back the mRNA sequence C into DNA.

•	Calculate the complementary sequences of DNA sequence A, sequence B, and sequence C.




```python
%pip install BioPython
import Bio
from Bio.Seq import Seq
```

```python
# Transcribe sequence A and sequence B into mRNA
seqA = Seq('ATGCGTTTATGACGTATCTGAAGTCTACTAGGGTGATCTGACTAA')
seqB = Seq('GTGTTAGTGCGCTAGTGCTGACTAGGTACAGTAGCTAGTTACTAG')
mRNA_A = seqA.transcribe()
mRNA_B = seqB.transcribe()
print(f"mRNA A = {mRNA_A}")
print(f"mRNA B = {mRNA_B}")
print(len(mRNA_A))
print(len(mRNA_B))
```

```python
# Combine 24 starting nucleotides of mRNA sequence A and 21 ending nucleotides of mRNA sequence B and set it as mRNA sequence C.
start_mRNA_A = mRNA_A[0:24]
End_mRNA_B = mRNA_B[24:]
print(len(start_mRNA_A))
print(len(End_mRNA_B))
mRNA_C = start_mRNA_A + End_mRNA_B
print(f"mRNA C = {mRNA_C}")
print(len(mRNA_C))
```

```python
# Translate the mRNA sequence A, sequence B, and sequence C into amino acids sequences.
aminoAcid_A = mRNA_A.translate()
aminoAcid_B = mRNA_B.translate()
aminoAcid_C = mRNA_C.translate()
print(f"Amino Acid of Sequence A {aminoAcid_A}")
print(f"Amino Acid of Sequence B {aminoAcid_B}")
print(f"Amino Acid of Sequence C {aminoAcid_C}")
```

```python
# Transcribe back the mRNA sequence C into DNA
# U menjadi T
seqC = ''
for a in range(0, len(mRNA_C)):
    if mRNA_C[a] == 'U':
        seqC += 'T'
        continue
    seqC += mRNA_C[a]

seqC = Seq(seqC)
print(f"mRNA C = {mRNA_C}")
print(f"DNA C = {seqC}")
    
```

```python
# Calculate the complementary sequences of DNA sequence A, sequence B, and sequence C.
seqA_antisense = seqA.complement()
seqB_antisense = seqB.complement()
seqC_antisense = seqC.complement()
print(f"Antisesense A: {seqA_antisense}")
print(f"Antisesense B: {seqB_antisense}")
print(f"Antisesense C: {seqC_antisense}")
```
